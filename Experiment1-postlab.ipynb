```python
# Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load Dataset
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
```

```python
# Display Data
print(df.head())
print(df.info())
print(df.describe())
print(df.isnull().sum())
```

```python
# Survival Count
sns.countplot(x='Survived', data=df)
plt.title("Survival Count(24EU01126)")
plt.show()
```

```python
# Gender Distribution
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title("Gender vs Survival()24EU01126")
plt.show()
```

```python
# Passenger Class vs Survival
sns.countplot(x='Pclass', hue='Survived', data=df)
plt.title("Pclass vs Survival(24EU01126)")
plt.show()
```
